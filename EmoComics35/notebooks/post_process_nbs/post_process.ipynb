```python
def parse_item(item):
    try:
        parsed = ast.literal_eval(item)
        if isinstance(parsed, list):
            return parsed
    except (SyntaxError, ValueError):
        pass
    return [item.strip("[]")]  # Handle '[Neutral]' case properly
```

```python
grounds = [[parse_item(item) for item in sublist] for sublist in grounds]
```

```python
bad_idx = []
for idx, (g, p) in enumerate(zip(grounds, predictions)):
    
    if len(g) != len(p):
        print(idx, len(g), len(p))
        bad_idx.append(idx)
```

```python
grounds = [elem for idx, elem in enumerate(grounds) if idx not in bad_idx]
predictions = [elem for idx, elem in enumerate(predictions) if idx not in bad_idx]
```

```python
len(grounds), len(predictions)
```

```python
grounds = [item for sublist in grounds for item in sublist]
predictions = [item for sublist in predictions for item in sublist]
```

```python
mlb = MultiLabelBinarizer()
```

```python
y_true_mhot = mlb.fit_transform(grounds)
y_pred_mhot = mlb.transform(predictions)
```

```python
print(classification_report(y_true_mhot, y_pred_mhot, target_names=mlb.classes_, digits=3))
```
